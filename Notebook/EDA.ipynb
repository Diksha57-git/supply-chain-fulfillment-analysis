### **UNIVARIANT ANALYSIS DESCRIPTION**

#### *CATEGORICAL COLUMN*
**1. Delivery Status**
  * "Late delivery" is by far the most frequent outcome, accounting for over 50% of all shipments.
  * Only about 18% of shipments are marked as "Shipping on time," indicating a significant bottleneck in regular scheduling.
  * "Advance shipping" is the second most common category at over 20%, while "Shipping canceled" remains minimal at under 5%.
    
More than half of the shipments are delivered late, while only around 18% are shipped on time. This indicates a significant delivery-performance issue and suggests that the company is struggling to consistently meet scheduled delivery commitments.

**2. Shipping Mode**
  * Standard Class is the most frequently used shipping mode.
  * Second Class is the second most commonly used shipping mode.
  * First Class accounts for approximately 15% of orders, while Same Day accounts for only around 5% of orders.
  * Overall, the company relies predominantly on Standard Class and Second Class shipping modes.
    
This indicates that the company primarily relies on Standard and Second Class shipping. Further analysis can be performed to determine whether the choice of shipping mode has an impact on delivery delays and overall delivery performance.

**3. Order Region**
  *  Central America has the highest share of orders, accounting for approximately 15.7% of total orders.
  *  Western Europe closely follows with around 15.0% of total orders.
  *  South America ranks third, contributing approximately 8.3% of total orders.
  *  The remaining regions individually account for less than 6% of total orders, indicating that order demand is relatively concentrated in Central        America and Western Europe.

**4. Market**
  * LATAM delivers the highest share of orders, accounting for approximately 28.57% of total orders.
  * Europe closely follows with around 27.83% of total orders,
  * Pacific Asia ranks third , contributing approximately 22.83% of total orders.
  * USCA and Africa contibutes rest of the total orders.

**5. Customer Segment**
  * The Consumer segment accounts for the largest share of orders, indicating that it is the company's primary customer base.
  * The Corporate segment follows with around 30.35% of total orders.
  * The Home Office segment have the least share of total orders, contributing approximately 17.8% share of orders.

**7. Order Status**
  * Complete is the most common order status, accounting for approximately 33% of all orders, indicating that only about one-third of orders have      successfully reached completion.
  * Pending Payment represents over 22% of orders, making it the second most frequent status and suggesting that payment completion may be a significant stage in the order lifecycle.
  * Processing, Pending, and Closed each account for around 10–12% of orders, indicating that a substantial proportion of orders remain in different stages of the order fulfillment process.
  * On Hold accounts for approximately 5.4% of orders, while Suspected Fraud and Canceled orders represent relatively small proportions of around 2% each.
  * Payment Review is the least common status, accounting for only around 1% of total orders.
    
While completed orders represent the largest individual category, a considerable proportion of orders remain in payment, pending, or processing stages. Further analysis could investigate whether these order statuses are associated with specific customer segments, regions, payment-related issues, or delivery performance.

### **BIVARIANT ANALYSIS DESCRIPTION**

* Central Africa has the highest proportion of late deliveries at approximately 58%, followed by South Asia and East Africa. However, the late-delivery rates among the top five regions are very similar, all exceeding 55%. This suggests that delivery delays are not limited to a single region and may indicate broader operational or logistics challenges.
  
* Western Europe may have one of the highest numbers of late deliveries because it has high order volume, but Central Africa has the highest rate of late deliveries.

* Shipping mode appears to have a significant impact on delivery performance. First Class has the worst performance, with over 95% of shipments delivered late, followed by Second Class with approximately 77% late deliveries. Same Day performs comparatively better, with nearly half of shipments delivered on time. Standard Class shows the most balanced delivery performance, with a relatively lower late-delivery rate and a high proportion of advance shipments.

* Actual shipping time varies significantly across shipping modes. Same Day shipping is the fastest, followed by First Class, while Second Class and Standard Class take approximately 4 days on average. The similar delivery duration of Second Class and Standard Class suggests limited differentiation between these two shipping options in terms of actual shipping speed.

* Late deliveries are the dominant delivery outcome across all major product categories. Leather, Men's Footwear, and Women's Apparel have the highest absolute number of late deliveries; however, their high order volumes may contribute to these counts. A normalized analysis of late-delivery rates is required to identify the categories with the worst actual delivery performance.

* Electronics has comparatively fewer late deliveries, but it also has a much lower number of total orders than most other displayed categories.

* All three customer segments have a very similar late-delivery rate of around 55%, indicating that delivery delays affect customers consistently across segments.
* Home Office has the slightly highest late-delivery rate, followed very closely by Consumer and Corporate, but the differences are minimal.
* Consumer customers have the highest proportion of advance shipping, compared with the other segments.
* Corporate customers show a comparatively higher proportion of on-time deliveries than the Consumer segment.
* Overall, customer segment does not appear to be a major differentiating factor in delivery performance, since late deliveries remain high across all segments.

* Actual shipping time generally increases with scheduled shipping time. However, shipments with the same scheduled delivery duration can have different actual shipping times, indicating variability in delivery performance. In particular, orders scheduled for 2 or 4 days show a wide range of actual shipping durations, suggesting that scheduled shipping time does not always accurately predict actual delivery time.Actual shipping time generally increases as the scheduled shipping duration increases. However, orders scheduled for 1 or 2 days take substantially longer than expected on average, suggesting potential delays in meeting delivery schedules. In contrast, orders scheduled for 3 days have an average actual shipping time close to the scheduled duration.
