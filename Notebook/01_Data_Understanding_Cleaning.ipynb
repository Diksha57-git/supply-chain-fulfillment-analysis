## **Key Findings**

* The dataset contains 0 duplicate rows, indicating each row represent unique transaction.
* Product description have 100% missing values thus it contains no description about the products.This column will be excluded for further analysis.
* Order Zipcode have 86% missing values,making it unsuitable for meaningful analysis. It will also be removed from the analytical dataset.
* Customer Zipcode and Customer Last Name contains only a few missing values,which have minimal impact on overall analysis.
* A large number of duplicate Order id and Customer id is observed, which is expected since a customer can place multiple order at the same time similarly, a single customer order may contain multiple products, resulting in multiple records with the same Order ID.

## **Data Cleaning**

```python
columns_to_drop= [
    "Product Description",
    "Order Zipcode",
    "Customer Email",
    "Customer Password",
    "Product Status"
]

df.drop(columns= columns_to_drop, inplace = True)
```
